## Experiment

Use the documented command to retrain on local WAV files. The implementation below is a mathematical smoke check, not a speech accuracy measurement.


```python
x=np.array([[-2,-2],[-3,-2],[2,2],[3,2]],dtype=float)
y=np.array([10,10,50,50])
model=MultinomialLogisticRegression(epochs=400).fit(x,y)
assert np.array_equal(model.predict(x),y)
model.loss_history[-1]
```
